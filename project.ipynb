The dataset was checked for common data quality issues such as missing values, duplicate rows, invalid values, and incorrect data types.

The following problems were identified:
1. Negative values in Income column
2. Negative values in LoanAmount column
3. Need to verify duplicates and missing values

```
df.duplicated().sum()
```

```
df.drop_duplicates(inplace=True)
```

```
df.isnull().sum()
```

```
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```
df = df[df['Income'] >= 0]
df = df[df['LoanAmount'] >= 0]
```

```
df.dtypes
```

```
df['Age'] = df['Age'].astype(int)
```

The following cleaning steps were performed on the dataset:

1. Imported dataset using pandas
2. Viewed dataset information using head(), info(), and describe()
3. Checked missing values using isnull()
4. Checked duplicate rows using duplicated()
5. Removed invalid negative values from Income and LoanAmount columns
6. Verified data types of all columns
7. Saved the cleaned dataset

```
df.describe()
```
